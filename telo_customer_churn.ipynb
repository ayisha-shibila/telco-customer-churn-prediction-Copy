```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
x = data.drop('Churn',axis=1)
y = data['Churn']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
import seaborn as sns
```

```python
from sklearn.preprocessing import StandardScaler
scalar_train = StandardScaler()
scaled_train_data = scalar_train.fit_transform(x_train)
x_train= pd.DataFrame(scaled_train_data,columns = x_train.columns)

scaler_test = StandardScaler()
scaled_test_data = scaler_test.fit_transform(x_test)
x_test= pd.DataFrame(scaled_test_data, columns = x_test.columns)
```

```python
from collections import Counter
```

```python
Counter(y_train)
```

```python
from imblearn.over_sampling import SMOTE

smt=SMOTE()
x_train_smt,y_train_smt=smt.fit_resample(x_train,y_train)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
le=LogisticRegression()
le.fit(x_train,y_train)
```

```python
y_pred1=le.predict(x_test)
```

```python
le.score(x_train,y_train)
```

```python
le.score(x_test,y_test)
```

```python
cm1= confusion_matrix(y_test,y_pred1)
```

```python
sns.heatmap(cm1,annot=True)
```

```python
accuracy_score(y_test,y_pred1)
```

```python
print(classification_report(y_test,y_pred1))
```
